```python
import os
import zipfile

# Unzipping the dataset into '/content/dataset'
zip_path = '/content/Cheating.v1i.yolov8.zip'
extract_path = '/content/dataset'

if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Dataset successfully unzipped!")
else:
    print(f"Error: {zip_path} not found. Please upload the dataset first.")
```

```python
# Training the model since dataset is now available
results = model.train(data=yaml_path, epochs=20, imgsz=640)
```

```python
import glob
import os
from ultralytics import YOLO

# Automatically find the latest training run's best.pt weights
runs = glob.glob('runs/detect/train*/weights/best.pt')
if runs:
    latest_run_weights = max(runs, key=os.path.getmtime)
    print(f"Loading model from: {latest_run_weights}")
    best_model = YOLO(latest_run_weights)
else:
    print("Error: Training weights not found. Please complete training first.")
```
