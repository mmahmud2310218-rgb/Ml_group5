```python
from ultralytics import YOLO

model = YOLO("yolov10n.pt")  # nano version (fast + good baseline)
```

```python
model.train(
    data="/content/drive/MyDrive/invigilator7/examCheating7/data2.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="runs/train",
    name="cheating_yolov10"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
results = model.predict(
    source="/content/drive/MyDrive/invigilator7/examCheating7/test/images",
    save=True,
    conf=0.25
)
```
